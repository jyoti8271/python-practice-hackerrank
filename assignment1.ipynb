```python
import math
num=int(input("enter the number"))
print(math.factorial(num))
```

```python
num=int(input("Enter the number"))

fact=1

for i in range(1,num+1):
    fact=fact*i
    
print(fact)
```

```python
def factorial(num):

    if num==0:
        return 1
    
    else:
        return num*factorial(num-1)
    
num=int(input("enter the number"))
print(factorial(num))
    
    
    
```
